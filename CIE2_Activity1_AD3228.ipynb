Dataset Description

The Students Performance dataset contains information about students' academic performance in mathematics, reading and writing. It also includes categorical information such as gender, parental education, lunch type and test preparation course.

The dataset contains 1000 records and 8 columns. It will be used to perform data cleaning, exploratory data analysis, categorical encoding and numerical feature scaling.

```python


# STEP 3: Basic Statistics
print("BASIC STATISTICS")
display(df.describe())

# STEP 4: Missing Value Analysis
print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nMISSING VALUE PERCENTAGE")
print((df.isnull().mean() * 100).round(2))

# STEP 5: Handle Missing Values
df_clean = df.copy()

num_cols = df_clean.select_dtypes(include=np.number).columns
cat_cols = df_clean.select_dtypes(include='object').columns

for col in num_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

for col in cat_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum().sum())

# STEP 6: Remove Duplicates and Fix Text
print("\nDuplicate rows:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

for col in cat_cols:
    df_clean[col] = df_clean[col].str.strip().str.lower()

print("Duplicates after cleaning:", df_clean.duplicated().sum())

# STEP 7: Exploratory Data Analysis (EDA)
print("\nGender Frequency:")
print(df_clean['gender'].value_counts())

# Visualization 1: Math Score Histogram
plt.figure(figsize=(7,4))
plt.hist(df_clean['math score'], bins=15, edgecolor='black')
plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.show()

# Visualization 2: Score Comparison
plt.figure(figsize=(7,4))
sns.boxplot(data=df_clean[['math score','reading score','writing score']])
plt.title("Comparison of Student Scores")
plt.ylabel("Marks")
plt.show()

print("\nEDA Observations:")
print("1. The histogram shows the distribution of mathematics scores.")
print("2. The box plot compares the spread and median of all three scores.")

# STEP 8: One-Hot Encoding
df_encoded = pd.get_dummies(
    df_clean,
    columns=cat_cols,
    drop_first=True,
    dtype=int
)

print("\nEncoded Dataset:")
display(df_encoded.head())

# STEP 9: Feature Scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Math score is kept unscaled as a possible future target.
scale_cols = ['reading score', 'writing score']

df_encoded[scale_cols] = scaler.fit_transform(
    df_encoded[scale_cols]
)

print("\nScaled Dataset:")
display(df_encoded.head())

# STEP 10: Save Prepared Dataset
print("\nFinal Dataset Shape:", df_encoded.shape)
print("Remaining Missing Values:", df_encoded.isnull().sum().sum())
print("Remaining Duplicates:", df_encoded.duplicated().sum())

df_clean.to_csv("Cleaned_StudentsPerformance.csv", index=False)
df_encoded.to_csv("Prepared_StudentsPerformance.csv", index=False)

print("\nBoth CSV files saved successfully!")

# Download prepared dataset
from google.colab import files
files.download("Prepared_StudentsPerformance.csv")
```
